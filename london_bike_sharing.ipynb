```python
season_mapping = {
   '0.0' : 'spring',
    '1.0' : 'summer',
    '2.0' : 'fall',
    '3.0' : 'winter'
}
df['season'] = df['season'].astype(str)
df['season'] = df['season'].map(season_mapping)
```

```python
weqther_mapping = {
    '1.0':'Clear',
    '2.0':'Scattered clouds',
    '3.0':'Broken clouds',
    '4.0':'Cloudy',
    '7.0':'Rain',
    '10.0':'Rain with thunderstorm',
    '26.0':'Snowfall'
}
df['weather'] = df['weather_code'].astype(str)
df['weather'] = df['weather'].map(weqther_mapping)
```
